```python
# using python list

import numpy as np
import time 

size = 1_000_0

l1 = list(range(size))
l2= list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end = time.time()
print(add[0:10])

print(end - start)

```

```python
# using python as numpy
import numpy as np
import time 

size = 1_000_0

l1 = np.array(list(range(size)))
l2 = np.array(list(range(size)))

start = time.time()
add = l1+l2
end = time.time()
print(add[0:10])

print(end - start)
```

```python
# here it is to check time only but of squaring
import numpy as np
import time

size = 1000000

list_square = list(range(size))
array_square = np.array(range(size))

start = time.time()

list_square = [x ** 2 for x in list_square]

end = time.time()

print("Python List Time:", end - start)

start = time.time()

array_square = array_square ** 2

end = time.time()

print("NumPy Array Time:", end - start)
```
