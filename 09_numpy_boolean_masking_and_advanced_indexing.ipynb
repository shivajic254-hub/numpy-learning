# Boolean Masking and Advanced Indexing

## Objective

Learn how to filter, search, and manipulate NumPy arrays using boolean masks, logical operations, and advanced indexing techniques.

## Reading Data from a File

```python
import numpy as np

filedata = np.genfromtxt("data.txt", delimiter=",")
filedata = filedata.astype("int32")

print(filedata)
```

```python
filedata > 50
```

```python
filedata[filedata > 50]
```

```python
np.any(filedata > 50, axis=0)
```

```python
np.all(filedata > 50, axis=0)
```
