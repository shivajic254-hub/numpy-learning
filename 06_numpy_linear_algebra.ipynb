# NumPy Linear Algebra

## Objective

Learn the basic linear algebra operations provided by NumPy.

### Topics Covered

- Matrix Multiplication
- Identity Matrix
- Determinant of a Matrix

```python
import numpy as np
```

```python
# Matrix Multiplication

a = np.ones((2,3))
print(a)

b = np.full((3,2),2)
print(b)

np.matmul(a,b)
```

```python
# Identity Matrix

c = np.identity(3)

c
```

```python
# Determinant

np.linalg.det(c)
```
