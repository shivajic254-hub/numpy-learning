# Python Lists vs NumPy Arrays

## Objective
Compare Python lists and NumPy arrays in terms of execution speed and understand why NumPy is preferred for numerical computing.

```python
# using python list

import numpy as np
import time 

size = 1_000_0

l1 = list(range(size))
l2= list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end = time.time()
print(add[0:10])

print(end - start)

```

## Key Takeaways

- Python lists require explicit iteration for element-wise operations.
- NumPy performs element-wise operations using vectorization.
- NumPy is generally faster for numerical computations.
- NumPy is widely used in data science and machine learning.
