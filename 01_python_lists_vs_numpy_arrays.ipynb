# Python Lists vs NumPy Arrays

## Objective

Compare Python lists and NumPy arrays in terms of execution speed and understand why NumPy is preferred for numerical computing.

```python
# Using Python Lists

import numpy as np
import time

size = 1_000_0

l1 = list(range(size))
l2 = list(range(size))

start = time.time()

add = [x + y for x, y in zip(l1, l2)]

end = time.time()

print(add[:10])
print(f"Python List Time: {end - start:.8f} seconds")
```

```python
# Using NumPy Arrays

import numpy as np
import time

size = 1_000_0

l1 = np.array(list(range(size)))
l2 = np.array(list(range(size)))

start = time.time()

add = l1 + l2

end = time.time()

print(add[:10])
print(f"NumPy Array Time: {end - start:.8f} seconds")
```

## Key Takeaways

- Python lists require explicit iteration for element-wise operations.
- NumPy performs element-wise operations using vectorization.
- NumPy generally executes numerical operations faster than Python lists.
- NumPy is widely used in data science, machine learning, and scientific computing.
