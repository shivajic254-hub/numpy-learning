# Creating NumPy Arrays

## Objective

Learn different ways to create NumPy arrays using built-in NumPy functions.

```python
## 1. Create an Array from a Python List
```

```python
import numpy as np
arr = np.array([1,2,3])
print(arr)
```

```python
## 2. Create an Array of Zeros
```

```python
np.zeros([3,4])
```

```python
np.ones([1,3])
```

```python
np.full((4,6),8)
```

```python
np.eye(5)
```

```python
np.eye(3)
```

```python
np.arange(1,100,9)
```

## Key Takeaways

- np.array() creates arrays from Python lists.
- np.zeros() creates arrays filled with zeros.
- np.ones() creates arrays filled with ones.
- np.full() fills an array with a specific value.
- np.eye() creates an identity matrix.
- np.arange() creates evenly spaced values.
