# NumPy Mathematical Operations

## Objective

Learn how to perform element-wise mathematical operations on NumPy arrays.

### Topics Covered
- Array Arithmetic
- Addition
- Multiplication
- Division
- Floor Division
- Power
- Array Addition
- Trigonometric Functions

```python
import numpy as np

a = np.array([1,2,3,4])
print(a)
```

```python
a + 2
```

```python
a * 2
```

```python
a // 2
```

```python
a ** 2
```

```python
b = np.array([9,8,7,6])

b
```
